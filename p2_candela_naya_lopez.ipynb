```python
import requests

# 1. GET a https://httpbin.org/html
response1 = requests.get('https://httpbin.org/html')
print("Respuesta 1:")
print("Código:", response1.status_code)
print("Cabeceras:", response1.headers)
print("Cuerpo:", response1.text)

# 2. GET a https://httpbin.org/html2 (Nota: Este endpoint podría no existir; si da 404, es normal, imprime lo que devuelva)
response2 = requests.get('https://httpbin.org/html2')
print("Respuesta 2:")
print("Código:", response2.status_code)
print("Cabeceras:", response2.headers)
print("Cuerpo:", response2.text)

# 3. GET a https://httpbin.org/forms/post (Esto devuelve un formulario HTML)
response3 = requests.get('https://httpbin.org/forms/post')
print("Respuesta 3:")
print("Código:", response3.status_code)
print("Cabeceras:", response3.headers)
print("Cuerpo:", response3.text)

# 4. POST a https://httpbin.org/post
# Primero, analiza el formulario del punto 3: tiene campos como 'custname', 'custtel', 'custemail', 'size', 'topping', 'delivery', 'comments'.
# Envía datos de ejemplo para todos.
data = {
    'custname': 'Nombre Ejemplo',
    'custtel': '123456789',
    'custemail': 'email@ejemplo.com',
    'size': 'medium',  # Opciones: small, medium, large
    'topping': ['cheese', 'mushroom'],  # Lista si es multiple
    'delivery': '18:00',
    'comments': 'Comentarios de prueba'
}
response4 = requests.post('https://httpbin.org/post', data=data)
print("Respuesta 4:")
print("Código:", response4.status_code)
print("Cabeceras:", response4.headers)
print("Cuerpo:", response4.text)
```

```python
import socket

# Crea socket TCP
sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Conecta a httpbin.org puerto 80
server_address = ('httpbin.org', 80)
sock.connect(server_address)

# Construye la petición HTTP
request = "GET /html HTTP/1.1\r\nHost: httpbin.org\r\nConnection: close\r\n\r\n"

# Envía la petición
sock.sendall(request.encode())

# Recibe la respuesta
response = b""
while True:
    data = sock.recv(4096)
    if not data:
        break
    response += data

# Cierra el socket
sock.close()

# Imprime la respuesta completa (decodificada)
print(response.decode())
```

```python
import requests

# 1. GET a http://httpbin.org/html (inseguro, podría redirigir)
response_http = requests.get('http://httpbin.org/html')
print("HTTP Inseguro:")
print("Código:", response_http.status_code)
print("Cabeceras:", response_http.headers)
print("Cuerpo:", response_http.text)

# 2. GET a https://httpbin.org/html (seguro)
response_https = requests.get('https://httpbin.org/html')
print("HTTPS Seguro:")
print("Código:", response_https.status_code)
print("Cabeceras:", response_https.headers)
print("Cuerpo:", response_https.text)
```

```python
import requests
from bs4 import BeautifulSoup

# 1. GET a la web de Práctica 1 (reemplaza con tu URL real)
url = 'https://naya-lopez-candela.mooo.com'  
response = requests.get(url)

# Parsea con BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# 2. Cuenta etiquetas <p>
num_parrafos = len(soup.find_all('p'))
print("Número de etiquetas <p>:", num_parrafos)

# 3. Muestra contenido de h1, h2 y pz
print("Contenido de h1:")
for h1 in soup.find_all('h1'):
    print(h1.text)

print("Contenido de h2:")
for h2 in soup.find_all('h2'):
    print(h2.text)

print("Contenido de p:")
for p in soup.find_all('p'):
    print(p.text)
```
